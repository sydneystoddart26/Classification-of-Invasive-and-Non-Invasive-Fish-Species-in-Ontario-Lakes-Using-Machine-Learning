# **Abstract**
Invasive fish species are a significant ecological threat to Ontario lakes, as they disrupt biodiversity native to the area and alter the aquatic ecosystems. When invasive species are identified, it is required to remove them, however, manual identification may be difficult sometimes for the general public who may not be familiar with all the possible fish species present within the region. This project aims to investigate the use of machine learning, particularly convolutional neural networks, to classify fish species as either invasive or native to the Ontario Lakes using image-based data. The dataset was compiled by collecting and preprocessing images from iNaturalist, using two different methods. The first method included simply resizing and labeling the data, while the second used a pre-trained YOLOv4 model to crop the fish. These methods were then individually trained and tested on a custom CNN model, and the data that used the first preprocessing method was then trained and tested on a pre-trained ResNet18 model. The results showed that both the custom CNN and the pre-trained ResNet18 model that were trained on the first dataset achieved about a 70% classification accuracy. However, training the custom CNN on the cropped data resulted in a lower classification accuracy, likely due to the incorrect cropping during YOLOv4 preprocessing. # **Experimental Setup**

**Dataset**

The data sources used for this project consist of open-source images from iNaturalist. The images were extracted into excel files that contained information relating to each image including:


*   Image ID
*   Image URL
*   Number of agreements with species identification
*   Number of disagreements with species identification

<br>

A python script was used to extract the images from their URLs and sent them into a RAW images folder which contained a separate folder for each species. Images with any disagreements on species identification along with any images with no agreements on species identification were filtered out in order to have the most accurate data possible. From here the images were then processed.  
<br>
The images were processed in two separate ways:  
<br>
**First method:**
The first method used to process the images was simply using a python script to resize each image to 224 by 224, as well as label it as either native or invasive. The script then split them into 70% training, 15% validation, and 15% testing. The images were then normalized during model training by applying the transform function which used a mean of [0.5, 0.5, 0.5] and a std of [0.5, 0.5, 0.5].  
<br>
**Second method:**
The second method involved a little more processing. As done in the first method, the images were resized and labeled the same. However, as some of the images contained quite a lot of noise, cropping was attempted on only the training images using a pre-trained YOLOv4 model in order to isolate and extract the fish from the background. The YOLOv4 model was sourced from the following public GitHub repository: https://github.com/tamim662/YOLO-Fish?tab=readme-ov-file. This project specifically used the configuration file yolov4.cfg which can be located inside the trained_on_merge folder which is contained in the models folder. After cropping was performed, the resulting images were than resized again to 224x224 and used in training with the non-cropped validation and testing images from the first method. As before they were then transofrmed during training and also used the same percentage data split.


<br>

**Machine Learning Techniques**

**Custom CNN:**
For this project a custom Convolutional Neural Network was built from scratch to perform binary classification between invasive and non-invasive species of fish in Ontario. Hyperparameter validation was performed and the best model was then used for testing. The exact details can be seen in the code block labeled Model towards the end of this report. Hyperparameter validation was done on the learning rate as well as the hidden layer size. Three learning rates were orginally tested, 0.01, 0.001, and 0.0001. However 0.01 was removed early in the process as the validation accuracy was not improving at all, no matter the hidden size value or the number of epochs.  
The hidden size hyperparameter was tested for 64 and 128 during validation testing with each of the different learning rates. The number of epochs was kept at 10 as there seemed to be no significant improvement when increasing this number. This was also combined with early stopping using a patience value of 3, in order to stop training when there were no improvements in the validation accuracy over consecutive epochs.
After validation the best model was then used for testing. This was done with both types of datasets from the preprocessing methods.
<br>  
**YOLO Prepocessing (Darknet):**  
YOLO, also known as You Only Look Once, is a real-time object detection system that uses a convolutional neural network in order to predict the bounding boxes and classes of objects in input images. For this project YOLOv4 was not directly trained, however it was used in order to crop out the fish from images as a preprocessing step for noise reduction. A pre-trained model YOLOv4 configuration, which had already learned to detect fish species was used as mentioned in the second method of data processing.  
This is an example of transfer learning, as the model has already been trained on a dataset and was then used for a different task. In this case that task was detecting and cropping fish from the original dataset images.
<br>   
The open-source neural network framework Darknet along with OpenCV was installed in order to run the YOLO model. For the data preprocessing step using the pre-trained YOLOv4 model, CPU computation was used. In order to run the model on my data images, I wrote two powershell scripts. The first script is used to detect the bounding boxes of identified fish for each image and store the information into a json file. These json files were all placed within a folder called detections which were then used in the second powershell script. The second script identified the coordinates of the bounding boxes, and then cropped a cloned copy of the original images, therefore removing a lot of the background noise.  
<br>
This pre-trained YOLO model was trained on underwater data. The model was chosen as some of the fish images from iNaturalist were underwater images, although there were also quite a few images that were fish above the water with individuals in the background either holding the fish or near it. This may have impacted the success rate of identifying fish within the input images.  

<br>

**Pre-trained ResNet18**  
Along with the custom CNN model, a pre-trained ResNet18 model was refined. ResNet18 was chosen due to its good balance between performance and speed, making it computationally feasable for the project as well as able to handle the needed image classification. The model is trained on a huge datase called ImageNet, which contains a large amount of every day images that helped the model learn basic features. For the fish classification task, early layers of the model were frozen and the final layer was retrained to recognize the two classes in this project. This was done as an attempt for the model to learn general features from the larger dataset and then be adapted to recognize invasive and native fish species.

```python
#class configuration
class Config:
  #data paths
  base_dir = "/content/data"
  labeled_dir = os.path.join(base_dir, "labeled_data")
  processed_dir = os.path.join(base_dir, "processed_images")

  train_dir = os.path.join(labeled_dir, "train")
  val_dir = os.path.join(labeled_dir, "val")
  test_dir = os.path.join(labeled_dir, "test")
  train_dir_processed = os.path.join(processed_dir, "train")

  output_dir = os.path.join(base_dir, "output")
  os.makedirs(output_dir, exist_ok=True)

  #model training settings
  num_classes = 2 #native and invasive species
  batch_size = 32
  num_epochs = [10]
  learning_rate = [0.001, 0.0001]
  hidden_sizes = [64, 128]
  model_name = "cnn"

config = Config()

print(f"Train path: {config.train_dir}")
print(f"Validation path: {config.val_dir}")
print(f"Test path: {config.test_dir}")
print(f"Processed images train path: {config.train_dir_processed}")
```

```python
#tranforming the images into tensors
def get_transforms():
  return transforms.Compose([transforms.RandomHorizontalFlip(),
                             transforms.RandomRotation(10),
                             transforms.Resize((224, 224)),
                             transforms.ToTensor(),
                             transforms.Normalize([0.5, 0.5, 0.5], [0.5, 0.5, 0.5])])

def get_transforms_for_visualization():
  return transforms.Compose([transforms.ToTensor()])
```

```python
#loading the datasets
def load_datasets(transform_function):
  transform = transform_function

  train_dataset = ImageFolder(config.train_dir, transform=transform)
  val_dataset = ImageFolder(config.val_dir, transform=transform)
  test_dataset = ImageFolder(config.test_dir, transform=transform)
  processed_train_dataset = ImageFolder(config.train_dir_processed, transform=transform)

  train_loader = DataLoader(train_dataset, batch_size=config.batch_size, shuffle=True)
  val_loader = DataLoader(val_dataset, batch_size=config.batch_size, shuffle=True)
  test_loader = DataLoader(test_dataset, batch_size=config.batch_size, shuffle=True)
  processed_train_loader = DataLoader(processed_train_dataset, batch_size=config.batch_size, shuffle=True)

  #classe names
  class_names = train_dataset.classes
  print(f"Class names: {class_names}")

  return train_loader, val_loader, test_loader, processed_train_loader, class_names

train_loader, val_loader, test_loader, processed_train_loader, class_names = load_datasets(get_transforms())
```

```python
#sample visualization function
t_loader, v_loader, te_loader, pro_loader, cnames = load_datasets(get_transforms_for_visualization())

def visualize_samples(loader, class_names, mean=(0.5, 0.5, 0.5), std=(0.5, 0.5, 0.5)):
  images, labels = next(iter(loader))
  plt.figure(figsize=(10, 10))

  for i in range(min(5, len(images))):
    img = images[i].clone()
    for im, m, s in zip (img, mean, std): #unnormalizing the images so that they look better
      im.mul(s).add_(m)

    img = img.permute(1, 2, 0).numpy()
    plt.subplot(1, 5, i+1)
    plt.imshow(img)
    plt.title(class_names[labels[i].item()])
    plt.axis('off')

  plt.show()

#example of a train visualization
visualize_samples(t_loader, cnames)
visualize_samples(pro_loader,cnames)
```

```python
#model function
def custom_cnn(num_classes):
  model = nn.Sequential(
      nn.Conv2d(3, 32, kernel_size=3, padding=1), #RGB are the 3 channel inputs
      nn.BatchNorm2d(32),
      nn.ReLU(),
      nn.MaxPool2d(2),

      nn.Conv2d(32, 64, kernel_size=3, padding=1),
      nn.BatchNorm2d(64),
      nn.ReLU(),
      nn.MaxPool2d(2),

      nn.Conv2d(64, 128, kernel_size=3, padding=1),
      nn.BatchNorm2d(128),
      nn.ReLU(),
      nn.MaxPool2d(2),

      nn.Flatten(),
      nn.Linear(128 * 28 * 28, 256),
      nn.ReLU(),
      nn.Dropout(0.3),
      nn.Linear(256, num_classes)
  )

  return model

untrained_model = custom_cnn(config.num_classes)
print(untrained_model)
```

```python
#training function
def train_epoch(model, train_loader, criterion, optimizer):
  model.train()
  total_loss = 0
  all_preds = []
  all_labels = []

  for images, labels in train_loader:
    #forward pass first
    outputs = model(images)
    loss = criterion(outputs, labels)

    #backpropagation
    optimizer.zero_grad()
    loss.backward()
    optimizer.step()

    #recording the loss and the predictions
    total_loss +=loss.item()
    _, preds = outputs.max(1)
    all_preds.extend(preds.tolist())
    all_labels.extend(labels.tolist())

  #averaging the loss and calculating accuracy
  avg_loss = total_loss / len(train_loader)
  accuracy = accuracy_score(all_labels, all_preds)

  return avg_loss, accuracy
```

```python
#validation function
def validate_epoch(model, val_loader, criterion):
  model.eval()
  total_loss = 0
  all_preds = []
  all_labels = []

  with torch.no_grad():
    for images, labels in val_loader:
      outputs = model(images)
      loss = criterion(outputs, labels)
      total_loss += loss.item()

      _, preds = outputs.max(1)
      all_preds.extend(preds.tolist())
      all_labels.extend(labels.tolist())

  #metrics
  avg_loss = total_loss / len(val_loader)
  accuracy = accuracy_score(all_labels, all_preds)
  precision = precision_score(all_labels, all_preds, average='weighted', zero_division=0)
  recall = recall_score(all_labels, all_preds, average='weighted', zero_division=0)
  f1 = f1_score(all_labels, all_preds, average='weighted', zero_division=0)

  metrics = {'val_loss': avg_loss, 'accuracy': accuracy, 'precision': precision,
             'recall': recall, 'f1_score': f1}

  return metrics
```

```python
#test function
def test_model(model, test_loader, criterion):
  model.eval()
  total_loss = 0
  all_preds = []
  all_labels = []

  with torch.no_grad():
    for images, labels in test_loader:
      outputs = model(images)
      loss = criterion(outputs, labels)
      total_loss += loss.item()

      _, preds = outputs.max(1)
      all_preds.extend(preds.tolist())
      all_labels.extend(labels.tolist())

  #metrics
  avg_loss = total_loss / len(test_loader)
  accuracy = accuracy_score(all_labels, all_preds)
  precision = precision_score(all_labels, all_preds, average='weighted', zero_division=0)
  recall = recall_score(all_labels, all_preds, average='weighted', zero_division=0)
  f1 = f1_score(all_labels, all_preds, average='weighted', zero_division=0)

  #tabulated results
  results = {
      'Loss' : [f"{avg_loss:.4f}"],
      'accuracy' : [f"{accuracy:.4f} "],
      'precision' : [f"{precision:.4f}"],
      'recall' : [f"{recall:.4f}"],
      'f1_score' : [f"{f1:.4f}"]
  }

  print(f"\nTest Results:")
  print(tabulate(pd.DataFrame(results), headers='keys', tablefmt='fancy_grid', showindex=False))

  return {'test_loss':avg_loss, 'accuracy':accuracy, 'precision':precision_score, 'recall':recall, 'f1_score':f1}, all_preds, all_labels
```

```python
#hyperparameter tuning
def tune_hyperparameters(learning_rates, hidden_sizes, train_loader, val_loader, num_epochs, patience=3):
  best_model = None
  best_accuracy = 0.0
  best_params = {}
  log = []

  for lr in learning_rates:
    for hidden_size in hidden_sizes:
      for epoch_count in num_epochs:
        print(f"Training with lr={lr}, hidden_size={hidden_size}")

        train_losses_per_run = []
        val_accuracies_per_run = []

        model = custom_cnn(config.num_classes)
        criterion = nn.CrossEntropyLoss()
        optimizer = optim.Adam(model.parameters(), lr=lr)

        no_improvement_epochs = 0
        best_model_for_run = None
        best_val_accu = 0.0

        for epoch in range(epoch_count):
          train_loss, train_accu = train_epoch(model, train_loader, criterion, optimizer)
          metrics = validate_epoch(model, val_loader, criterion)
          val_accu = metrics['accuracy']
          train_losses_per_run.append(train_loss)
          val_accuracies_per_run.append(val_accu)
          print(f"Epoch {epoch+1} | Train Loss: {train_loss:.4f} | Val accuracy: {val_accu:.4f}")

          if val_accu > best_val_accu:
            best_val_accu = val_accu
            best_model_for_run = copy.deepcopy(model)
            no_improvement_epochs = 0

          else:
            no_improvement_epochs += 1

            if no_improvement_epochs >= patience:
              print("Early stopping")
              break

        log.append({'lr':lr, 'hidden_size': hidden_size, 'val_accu' : best_val_accu,
                    'train_losses' : train_losses_per_run, 'val_accuracies' : val_accuracies_per_run})

        if best_val_accu > best_accuracy:
          best_accuracy = best_val_accu
          best_model = best_model_for_run
          best_params = {'lr' : lr, 'hidden_size' : hidden_size}

  print(f"Best parameters: {best_params}")

  #plotting the training curve for each combination of hyperparameters
  for entry in log:
    plt.figure(figsize=(10,6))
    plt.plot(entry['train_losses'], label='Training Loss')
    plt.plot(entry['val_accuracies'], label='Validation Accuracy')
    plt.title(f"LR: {entry['lr']} | Hidden Size: {entry['hidden_size']}")
    plt.xlabel("Epoch"), plt.ylabel("Value")
    plt.legend()
    plt.show()

  #hyperparameter table
  df = pd.DataFrame(log)
  df = df[['lr', 'hidden_size', 'val_accu']]
  df.columns = ['Learning Rate', 'Hidden Size', 'Validation Accuracy']
  #rounding validation accuracy to 4 decimal places
  df['Validation Accuracy'] = df['Validation Accuracy'].apply(lambda x: f"{x:.4f}")
  print("\nHyperparameter Tuning Summary:")
  print(tabulate(df, headers='keys', tablefmt='fancy_grid'))

  return best_model
```

```python
#confusion matrix and metrics report
def plot_confusion_matrix(all_labels, all_preds):
  conf_matrix = confusion_matrix(all_labels, all_preds)
  sns.heatmap(conf_matrix, annot=True, fmt="d", cmap="Blues",
              xticklabels=["Invasive", "Native"], yticklabels=["Invasive", "Native"])
  plt.xlabel("Predicted")
  plt.ylabel("True")
  plt.title("Confusion Matrix")
  plt.show()

#classification table report
def classification_report_table(all_labels, all_preds, class_names):
  report = classification_report(all_labels, all_preds, target_names=class_names, output_dict=True)
  df = pd.DataFrame(report).transpose()
  print(tabulate(df[['precision', 'recall', 'f1-score', 'support']], headers='keys', tablefmt='fancy_grid'))
```

```python
#showing example predictions function
def show_example_predictions(model, data_loader, class_names, num_images=4):
  model.eval()
  images_shown = 0 #image counter
  plt.figure(figsize=(15,10))

  with torch.no_grad():
    #getting the model prediction
    for images, labels in data_loader:
      outputs = model(images)
      _, predicts = torch.max(outputs, 1)

      for i in range(images.size(0)): #looping through the batch
        if images_shown >= num_images: #stopping when 4 images have been shown
          return

        plt.subplot(2, 2, images_shown + 1) #subplot for the image
        plt.imshow(np.transpose(images[i].numpy(), (1, 2, 0)))
        #adding the labels
        plt.title(f"True: {class_names[labels[i]]}\nPred: {class_names[predicts[i]]}")
        plt.axis('off')
        images_shown += 1
```
